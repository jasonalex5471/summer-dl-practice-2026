```python
import torch
x = torch.arange(40,dtype = torch.float32).reshape(5,8)
print(x)

x_sum_axis0 = x.sum(axis = 0)   #降维求和
print(x_sum_axis0)
print(x_sum_axis0.shape)
x_sum_axis1 = x.sum(axis = 1)
print(x_sum_axis1)
print(x_sum_axis1.shape)

x_sum_axis0_keepdims = x.sum(axis=0, keepdims=True) #非降维求和
print(x_sum_axis0_keepdims)
print(x_sum_axis0_keepdims.shape)
x_sum_axis1_keepdims = x.sum(axis=1, keepdims=True)
print(x_sum_axis1_keepdims)
print(x_sum_axis1_keepdims.shape)

print(x.cumsum(axis=0))     #cumsum函数不会沿任何轴降低维度
print(x.cumsum(axis=0).shape)
```
