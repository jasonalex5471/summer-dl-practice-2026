```python
import torch
A = torch.arange(20, dtype=torch.float32).reshape(5,4)      #矩阵-向量积的计算
print(A)
print(A.shape)

x = torch.tensor([1.,2.,5.,1.])

print(x)
print(x.shape)

print(torch.mv(A,x))
```

```python
A = torch.ones(20, dtype=torch.float32).reshape(4,5)

B = torch.ones(20, dtype=torch.float32).reshape(5,4)

print(A)
print(B)

print(A @ B)                    #矩阵-矩阵乘法
print((A @ B).shape)

print(torch.mm(A,B))
print((torch.mm(A,B)).shape)
```

```python
%matplotlib inline
import torch
import matplotlib.pyplot as plt
from matplotlib_inline import backend_inline

#用matplotlib来演示求极限

# 1. 设置渲染格式为 SVG（保证生成的图像无限放大不糊）
backend_inline.set_matplotlib_formats('svg')

# 2. 定义目标函数 f(x) = 3*x^2 - 4*x
def f(x):
    return 3 * x**2 - 4 * x

# 3. 构建 X 轴的数据（使用 PyTorch Tensor 替换 np.arange）
# 生成从 0 到 3，步长为 0.1 的张量
x = torch.arange(0, 3, 0.1)

# 4. 计算 Y 轴数据
# y1: 函数 f(x) 的曲线值
# y2: x = 1 处的切线方程 y = 2x - 3
y1 = f(x)
y2 = 2 * x - 3

# 5. 原生 Matplotlib 面向对象绘图流程
# 创建画布和坐标轴对象 (figsize 设置图标宽和高)
fig, ax = plt.subplots(figsize=(4.5, 3.0))

# 在坐标轴上绘制两条曲线
# '-' 表示实线，'--' 表示虚线
ax.plot(x.numpy(), y1.numpy(), '-', label='f(x)')
ax.plot(x.numpy(), y2.numpy(), '--', label='Tangent line (x=1)')

# 配置坐标轴属性（原图中 set_axes 函数的核心逻辑）
ax.set_xlabel('x')                            # 设置 X 轴标签
ax.set_ylabel('f(x)')                         # 设置 Y 轴标签
ax.set_xlim([0, 3])                           # 设置 X 轴范围
ax.set_ylim([-3, 13])                         # 设置 Y 轴范围
ax.legend(loc='upper left')                   # 显示图例并放在左上角
ax.grid(True)                                 # 开启背景网格线

# 6. 显示图表
plt.show()
```
