```python
import torch
X = torch.arange(12,dtype=torch.float32).reshape(3,4)
print(X)
print(X[-1])
print(X[1:3])
X[1,2] = 9
print(X)
X[0:2,:] = 12
print(X)
```

```python
Y = torch.ones(3,4)
X = torch.zeros(3,4)
Z = torch.zeros_like(Y)
print(id(Z))
Z[:] = X + Y    #节约内存
print(id(Z))
```
